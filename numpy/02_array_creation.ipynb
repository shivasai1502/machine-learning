# Array Creation

NumPy provides many ways to create arrays for different purposes.

## Common Array Creation Functions
- `np.array()`: Create an array from a Python list or tuple
- `np.arange()`: Create arrays with regularly incrementing values
- `np.linspace()`: Create arrays with a specified number of values between a start and stop value
- `np.zeros()`, `np.ones()`: Create arrays filled with zeros or ones
- `np.eye()`: Create an identity matrix
- `np.diag()`: Create a diagonal matrix
- `np.vander()`: Create a Vandermonde matrix

---

## Why Array Creation?

- Creating arrays efficiently is essential for scientific computing and data analysis.
- NumPy provides many functions to create arrays for different needs: sequences, filled arrays, identity matrices, and more.
- Understanding these functions helps you write faster and more readable code.


````python
import numpy as np
````

````python
# np.arange: Create a 1D array with values from 0 to 10 (step 2)
arr = np.arange(0, 11, 2)
print('arange:', arr)
````

````python
# np.linspace: Create 10 values between 0 and 5 (inclusive)
arr = np.linspace(0, 5, 10)
print('linspace:', arr)
````

````python
# np.zeros: Create an array filled with zeros
zeros_arr = np.zeros((2, 3))
print('zeros:', zeros_arr)
````

````python
# np.ones: Create an array filled with ones
ones_arr = np.ones((2, 3))
print('ones:', ones_arr)
````

````python
# np.eye: Identity matrix
eye_arr = np.eye(3)
print('eye:', eye_arr)
````

````python
# np.diag: Diagonal matrix
diag_arr = np.diag([1, 2, 3, 4])
print('diag:', diag_arr)
````

````python
# np.vander: Vandermonde matrix
vander_arr = np.vander([1, 2, 3, 4], 3)
print('vander:', vander_arr)
````

## Common Errors

- **Shape mismatch:** Passing wrong shape arguments to creation functions.
- **Type errors:** Using unsupported data types.
- **Confusing `arange` and `linspace`:** `arange` uses step, `linspace` uses number of elements.

**Example:**
```python
# np.zeros(2, 3)  # TypeError: shape must be a tuple
np.zeros((2, 3))  # Correct
```
