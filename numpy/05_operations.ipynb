# Array Operations

NumPy supports a wide range of operations on arrays, including arithmetic, vectorized math, comparisons, broadcasting, and filtering.

## Arithmetic Operations
- Addition, subtraction, multiplication, division, and more can be performed element-wise.

**Syntax:**
```python
arr1 + arr2
arr1 - arr2
arr1 * arr2
arr1 / arr2
arr1 ** arr2  # exponentiation
```

## Vectorized Functions
- NumPy provides fast math functions that operate on arrays.
- Examples: `np.sqrt`, `np.exp`, `np.log`, `np.sin`, `np.cos`, etc.

## Comparison & Boolean Operations
- Compare arrays or with scalars: `==`, `!=`, `>`, `<`, `>=`, `<=`
- Boolean indexing: filter elements based on conditions

## Broadcasting
- Allows operations on arrays of different shapes by 'stretching' dimensions.

## Why Learn Array Operations?

Array operations are the foundation of data analysis and scientific computing with NumPy. They allow you to:
- Perform fast, element-wise calculations on large datasets.
- Write concise and readable code (no need for explicit loops).
- Leverage broadcasting to work with arrays of different shapes.
- Apply mathematical functions efficiently to entire datasets.

Understanding these operations will help you write code that is both efficient and easy to understand.

````python
import numpy as np
````

````python
# Arithmetic Operations: Element-wise Math

arr1 = np.array([10, 20, 30, 40])
arr2 = np.array([1, 2, 3, 4])

# Addition
print('Addition:', arr1 + arr2)

# Subtraction
print('Subtraction:', arr1 - arr2)

# Multiplication
print('Multiplication:', arr1 * arr2)

# Division
print('Division:', arr1 / arr2)

# Addition with scalar
print('Addition with scalar:', arr1 + 5)
````

### Arithmetic Operations

You can perform element-wise arithmetic operations on NumPy arrays. This means that operations like addition, subtraction, multiplication, and division are applied to each element in the arrays.

**Example:**

````python
# Vectorized Math Functions
import numpy as np

arr = np.array([1, 2, 3, 4, 5])

# Square root
print('Square root:', np.sqrt(arr))

# Exponential
print('Exponential:', np.exp(arr))

# Logarithm
print('Logarithm:', np.log(arr))

# Sine
print('Sine:', np.sin(arr))

# Cosine
print('Cosine:', np.cos(arr))
````

### Vectorized Math Functions

NumPy provides fast mathematical functions that operate on entire arrays at once. These are called vectorized functions and include operations like square root, exponential, logarithm, sine, and cosine.

**Example:**

````python
# Element-wise Power
import numpy as np

arr1 = np.array([1, 2, 3, 4])
arr2 = np.array([5, 6, 7, 8])

# Each element in arr1 is raised to the power of the corresponding element in arr2
print('Power:', arr1 ** arr2)
````

### Element-wise Power

You can raise each element of an array to the power of the corresponding element in another array using the `**` operator.

**Example:**

````python
# Comparison and Boolean Indexing
import numpy as np

arr1 = np.array([10, 20, 30, 40])
arr2 = np.array([15, 20, 25, 35])

# Element-wise comparison
print('Equal:', arr1 == arr2)
print('Less than 25:', arr1 < 25)

# Modify elements using boolean indexing
arr1[arr1 < 25] = 0
print('After filtering:', arr1)
````

### Comparison and Boolean Indexing

You can compare arrays element-wise and use boolean indexing to filter or modify elements based on conditions.

**Example:**

````python
# Broadcasting Example
import numpy as np

arr = np.array([[1, 2, 3], [4, 5, 6]])
scalar = 10

# The scalar is broadcasted to each element of the array
result = arr + scalar
print('Broadcasting result:\n', result)
````

### Broadcasting

Broadcasting allows NumPy to perform operations on arrays of different shapes by automatically expanding their dimensions when possible.

## Common Errors & Pitfalls

1. **Shape Mismatch:**
   - Trying to perform operations on arrays with incompatible shapes will raise a `ValueError`.
   - Example:
     ```python
     arr1 = np.array([1, 2, 3])
     arr2 = np.array([1, 2])
     arr1 + arr2  # ValueError: operands could not be broadcast together
     ```
2. **Integer Division:**
   - In Python 3, `/` always returns a float. Use `//` for integer division if needed.
   - Example:
     ```python
     arr = np.array([5, 10, 15])
     print(arr / 2)   # [2.5 5.  7.5]
     print(arr // 2)  # [2 5 7]
     ```
3. **In-place Modification with Boolean Indexing:**
   - Be careful when modifying arrays in place using boolean indexing, as it can lead to unexpected results if not used properly.
   - Example:
     ```python
     arr = np.array([1, 2, 3, 4])
     arr[arr > 2] = 0  # [1 2 0 0]
     ```
4. **Broadcasting Confusion:**
   - Broadcasting rules can be tricky. Make sure you understand how dimensions are matched.
   - Example:
     ```python
     arr = np.array([[1, 2, 3], [4, 5, 6]])
     vec = np.array([1, 2])
     arr + vec  # ValueError: operands could not be broadcast together
     ```
