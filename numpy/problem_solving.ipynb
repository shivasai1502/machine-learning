### 2. Create a 3x3 matrix with values ranging from 0 to 8. Replace all values greater than 4 with -1.

```python
arr = np.random.randint(0,8, size=(3,3))
print(arr)
arr[arr > 4] = -1
print(arr)
```

### 3. Create two arrays: `a = [1, 2, 3, 4, 5]` and `b = [6, 7, 8, 9, 10]`. Stack them vertically and then horizontally.

```python
a=np.array([1,2,3,4,5])
b=np.array([6,7,8,9,10])

c=np.vstack((a,b))
print(c)

d=np.hstack((a,b))
print(d)
```

### 4. Create a 5x5 matrix with random values and find the minimum and maximum values.

### 5. Create a 10x10 array with random values and normalize it (scale values between 0 and 1).

```python
a = np.random.randint(1, 100, size=(10,10))

print(a)
#normalize the array
a_normalized = (a - np.min(a)) / (np.max(a) - np.min(a))
print(a_normalized)
```

### 7. Create a 4x4 matrix and extract the border elements (first/last row and first/last column).

```python
a=np.random.randint(1,10,size=(4,4))
print(a)

border = np.concatenate([a[0],a[-1],a[:,0],a[:,-1]])
print(border)
border.sort()
print(border)
```

### 8. Find the most frequent value in the array: `[1, 2, 3, 4, 4, 4, 5, 5, 6, 6, 6, 6, 7]`

```python

arr = np.array([1,2,3,4,4,4,4,5,6,6,7,7,7,7,7,8,9,9,10,10,10])
val, count = np.unique(arr, return_counts=True)

print(val)
print(count)

index = np.argmax(count)
print(f'Most frequent value is {val[index]} with count {count[index]}')
```

### 9. Create two arrays of shape (3, 4) and (4, 5) and perform matrix multiplication.

```python
a=np.random.randint(1,100,size=(3,4))
b=np.random.randint(1,100,size=(4,5))

c = np.dot(a,b)
print(c)

d=np.matmul(a,b)
print(d)
```

### 10. Given a 2D array, swap two random rows and two random columns.

```python
a=np.random.randint(1,100,size=(2,2))

b=a.T
print(b)

#swap two randowm rows and two random columns
a=np.random.randint(1,100,size=(4,5))
print("Original array:")
print(a)
# Swap two random rows
print(a.shape[0])
row1, row2 = np.random.choice(a.shape[0], size=2, replace=False)
a[[row1, row2], :] = a[[row2, row1], :]
print("Array after swapping rows:")
print(a)
# Swap two random columns
col1, col2 = np.random.choice(a.shape[1], size=2, replace=False)
a[:, [col1, col2]] = a[:, [col2, col1]]
print("Array after swapping columns:")
print(a)
```

### 11. Create a 10x10 array with random integers from 1 to 100. Find all rows where the sum of elements is greater than 500.

```python
a = np.random.randint(1, 100, size=(10,10))

print(a[np.sum(a, axis=1) > 500])  # Rows where sum is greater than 500
```

### 12. Compute the Euclidean distance between two arrays: `a = [1, 2, 3]` and `b = [4, 5, 6]`

```python
a=np.array([1,2,3,4,5])
b=np.array([6,7,8,9,10])

c = np.linalg.norm(a - b)
print(c)
```
