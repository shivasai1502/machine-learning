# Random Numbers in NumPy

NumPy provides several functions to generate random numbers and arrays. These are useful for simulations, testing, and initializing data.

## Common Random Functions
- `np.random.rand()`: Uniform distribution over [0, 1)
- `np.random.randint()`: Random integers from a range
- `np.random.randn()`: Standard normal distribution (mean 0, std 1)
- `np.random.uniform()`: Uniform distribution over a custom interval

---

## Why Random Numbers?

- Random numbers are essential for simulations, testing, sampling, and initializing machine learning models.
- NumPy provides fast and flexible random number generation for arrays of any shape.


## About `np.random.rand`


- **Purpose:** Generates random numbers from a uniform distribution over [0, 1).
- **Shape:** Pass one or more integers as arguments to specify the shape of the output array. For example, `np.random.rand(2, 3)` creates a 2x3 array.
- **Data type:** The output is always of float type.
- **Reproducibility:** You can set a random seed using `np.random.seed()` to get repeatable results.
- **Usage:** Useful for simulations, initializing weights in machine learning, and random sampling.

**Example:**
```python
np.random.seed(42)  # Set the seed for reproducibility
arr = np.random.rand(2, 3)
print(arr)
```

````python
import numpy as np
````

````python
# np.random.rand: Uniform random numbers between 0 and 1
arr = np.random.rand(3, 4)
print('rand:', arr)
````

````python
# np.random.randint: Random integers between 1 and 9 (inclusive of 1, exclusive of 10)
arr = np.random.randint(1, 10, (3, 4))
print('randint:', arr)
````

````python
# np.random.randn: Standard normal distribution
arr = np.random.randn(3, 4)
print('randn:', arr)
````

````python
# np.random.uniform: Uniform random numbers between 0 and 42
arr = np.random.uniform(0, 42, size=(3, 4))
print('uniform:', arr)
````

## Common Errors

- **Shape mismatch:** Passing wrong shape arguments to random functions.
- **Type errors:** Using unsupported data types or wrong argument order.
- **Forgetting to set a seed:** Results are not reproducible unless you use `np.random.seed()`.

**Example:**
```python
# np.random.randint(5, size=2, 3)  # TypeError: size must be a tuple or int
np.random.randint(5, size=(2, 3))  # Correct
```
