# NumPy Basics

NumPy is a powerful Python library for numerical computing. It provides support for large, multi-dimensional arrays and matrices, along with a collection of mathematical functions to operate on them efficiently.

## What is a NumPy Array?
A NumPy array ("ndarray") is a grid of values, all of the same type, indexed by a tuple of nonnegative integers.

### Creating a NumPy Array
**Syntax:**
```python
import numpy as np
arr = np.array([1, 2, 3])
```

### Array Properties
- `shape`: Dimensions of the array
- `dtype`: Data type of the array elements
- `size`: Total number of elements
- `ndim`: Number of dimensions
- `itemsize`: Size in bytes of each element

---

## Why NumPy Arrays?

- NumPy arrays are faster and use less memory than Python lists.
- They support vectorized operations (apply operations to all elements at once).
- Useful for mathematical and scientific computing.

**Example:**
- Adding two lists vs. two arrays.


````python
# Python list vs NumPy array
py_list1 = [1, 2, 3]
py_list2 = [4, 5, 6]
print('List addition:', py_list1 + py_list2)  # Concatenates lists

import numpy as np
np_arr1 = np.array([1, 2, 3])
np_arr2 = np.array([4, 5, 6])
print('Array addition:', np_arr1 + np_arr2)  # Element-wise addition
````

````python
import numpy as np
````

````python
# Create a 1D array from a Python list
arr = np.array([1, 2, 3, 4, 5])
print('1D array:', arr)
````

````python
# Create a 2D array (matrix)
arr2d = np.array([[1, 2, 3], [4, 5, 6]])
print('2D array:\n', arr2d)
````

````python
# Array properties for 1D array
print('Array Shape:', arr.shape)
print('Array Data Type:', arr.dtype)
print('Array Size:', arr.size)
print('Array Dimensions:', arr.ndim)
print('Array Item Size (bytes):', arr.itemsize)
````

````python
# Array properties for 2D array
print('2D Array Shape:', arr2d.shape)
print('2D Array Data Type:', arr2d.dtype)
print('2D Array Size:', arr2d.size)
print('2D Array Dimensions:', arr2d.ndim)
print('2D Array Item Size (bytes):', arr2d.itemsize)
````

## Reshaping Arrays

You can change the shape of an array using the `reshape` method.

**Example:**

````python
# Reshape a 1D array to 2D
arr = np.array([1, 2, 3, 4, 5, 6])
reshaped = arr.reshape(2, 3)
print('Original:', arr)
print('Reshaped (2x3):\n', reshaped)
````

## Common Errors

- **Shape mismatch:** Trying to reshape to an incompatible shape or operate on arrays of different shapes.
- **Type errors:** Mixing lists and arrays in operations.

**Example:**
```python
arr = np.array([1, 2, 3])
# arr.reshape(2, 2)  # ValueError: cannot reshape array of size 3 into shape (2,2)
```
