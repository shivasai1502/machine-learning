# Pandas Basics - Data Manipulation for Machine Learning

Pandas is essential for data preprocessing and exploration.

```python
import pandas as pd
import numpy as np
```

## 1. Creating DataFrames

```python
# From dictionary
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Age': [25, 30, 35, 28, 32],
    'Score': [85.5, 92.0, 78.5, 88.0, 95.5],
    'City': ['NYC', 'LA', 'Chicago', 'NYC', 'LA']
}

df = pd.DataFrame(data)
print("DataFrame:")
print(df)
print(f"\nShape: {df.shape}")
print(f"\nData types:\n{df.dtypes}\n")
```

## 2. Basic Operations

```python
print("First 3 rows:")
print(df.head(3))
print("\nBasic statistics:")
print(df.describe())
print("\nColumn names:", df.columns.tolist())
print("\nInfo:")
print(df.info())
```

## 3. Selecting Data

```python
# Select column
print("Ages:")
print(df['Age'])
```

```python
# Select multiple columns
print("Names and Scores:")
print(df[['Name', 'Score']])
```

```python
# Select rows by condition
print("People older than 28:")
print(df[df['Age'] > 28])
```

```python
# Multiple conditions
print("People older than 28 with score > 90:")
print(df[(df['Age'] > 28) & (df['Score'] > 90)])
```

## 4. Data Manipulation

```python
# Add new column
df['Score_Normalized'] = (df['Score'] - df['Score'].mean()) / df['Score'].std()
print("With normalized score:")
print(df)
```

```python
# Group by operations
print("Average age by city:")
print(df.groupby('City')['Age'].mean())
```

```python
print("Statistics by city:")
print(df.groupby('City').agg({
    'Age': 'mean',
    'Score': ['mean', 'max', 'min']
}))
```

## 5. Handling Missing Data

```python
# Create data with missing values
data_missing = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [5, np.nan, np.nan, 8, 9],
    'C': [10, 11, 12, 13, 14]
}

df_missing = pd.DataFrame(data_missing)
print("Data with missing values:")
print(df_missing)

print("\nMissing values count:")
print(df_missing.isnull().sum())
```

```python
print("Fill missing with mean:")
df_filled = df_missing.fillna(df_missing.mean())
print(df_filled)
```

```python
print("Drop rows with missing values:")
df_dropped = df_missing.dropna()
print(df_dropped)
```

## 6. Sorting and Ranking

```python
print("Sort by Score (descending):")
print(df.sort_values('Score', ascending=False))
```

```python
print("Rank scores:")
df['Rank'] = df['Score'].rank(ascending=False)
print(df[['Name', 'Score', 'Rank']])
```
